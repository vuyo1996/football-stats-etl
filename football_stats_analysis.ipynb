```python
import pandas as pd
```

```python
pd.set_option("display.max_rows", 10)
pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../Datasets/football.csv")
```

```python
df = df.reset_index(drop=True)
df["player_id"] = df.index + 1

cols = ["player_id"] + [col for col in df.columns if col != "player_id"]
df = df[cols]
```

```python
df.shape
```

```python
df.head()
```

```python
# Standardize column names
df.columns = df.columns.str.lower().str.strip().str.replace(" ", "_")
```

```python
df["age"] = df["age"].astype("Int64")
```

```python
count_cols = [
    "jersey_number",
    "headed_goals",
    "goals_with_right_foot",
    "goals_with_left_foot",
    "penalties_scored",
    "freekicks_scored",
    "shots",
    "shots_on_target",
    "hit_woodwork",
    "big_chances_missed",
    "clean_sheets",
    "goals_conceded",
    "tackles",
    "last_man_tackles",
    "blocked_shots",
    "interceptions",
    "clearances",
    "headed_clearance",
    "clearances_off_line",
    "recoveries",
    "duels_won",
    "duels_lost",
    "aerial_battles_won",
    "aerial_battles_lost",
    "own_goals",
    "errors_leading_to_goal",
    "assists",
    "passes",
    "big_chances_created",
    "crosses",
    "through_balls",
    "accurate_long_balls",
    "saves",
    "penalties_saved",
    "punches",
    "high_claims",
    "catches",
    "sweeper_clearances",
    "throw_outs",
    "goal_kicks",
    "yellow_cards",
    "red_cards",
    "fouls",
    "offsides"
]

for col in count_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce").astype("Int64")
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df
```

```python
# Convert percentage columns to numeric
percent_cols = ["shooting_accuracy_%",
                "tackle_success_%",
                "cross_accuracy_%"]

for col in percent_cols:
    if col in df.columns:
        df[col] = df[col].astype(str).str.replace("%", "", regex=False)
        df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df = df.rename(columns={
    "shooting_accuracy_%": "shooting_accuracy",
    "tackle_success_%": "tackle_success",
    "cross_accuracy_%": "cross_accuracy",
    "successful_50/50s": "successful_50_50s"
})
```

```python
df.groupby("position").agg(
    player_count=("name", "count"),
    saves_missing=("saves", lambda x: x.isna().sum()),
    goals_per_match_missing=("goals_per_match", lambda x: x.isna().sum()),
    clean_sheets_missing=("clean_sheets", lambda x: x.isna().sum())
)
```

```python
players = df[["player_id",
              "name",
              "club",
              "position",
              "nationality",
              "age",
              "appearances",
              "wins",
              "losses"]]
```

```python
goalkeeper_stats = df[df["position"] == "Goalkeeper"][["player_id",
                                                       "club",
                                                       "saves",
                                                       "penalties_saved",
                                                       "punches",
                                                       "high_claims",
                                                       "catches",
                                                       "sweeper_clearances",
                                                       "throw_outs",
                                                       "goal_kicks",
                                                       "clean_sheets"]]
```

```python
goalkeeper_stats.isnull().sum().sum()
```

```python
attacking_stats = df[df["position"].isin(["Forward", "Midfielder"])][["player_id",
                                                                      "club",
                                                                      "goals",
                                                                      "goals_per_match",
                                                                      "shots",
                                                                      "shots_on_target",
                                                                      "shooting_accuracy"]]
```

```python
# Remove any row where ANY of these columns has a missing value (NaN)
attacking_stats = attacking_stats.dropna(subset=["goals_per_match",
                                                 "shots",
                                                 "shots_on_target",
                                                 "shooting_accuracy"])
```

```python
attacking_stats
```

```python
attacking_stats.isnull().sum()
```

```python
defensive_stats = df[df["position"].isin(["Defender", "Midfielder"])][["player_id",
                                                                       "club",
                                                                       "tackles",
                                                                       "tackle_success",
                                                                       "last_man_tackles",
                                                                       "blocked_shots",
                                                                       "interceptions",
                                                                       "clearances",
                                                                       "headed_clearance",
                                                                       "clearances_off_line",
                                                                       "recoveries",
                                                                       "duels_won",
                                                                       "duels_lost",
                                                                       "successful_50_50s",
                                                                       "aerial_battles_won",
                                                                       "aerial_battles_lost",
                                                                       "own_goals",
                                                                       "errors_leading_to_goal"]]
```

```python
defensive_stats["own_goals"] = defensive_stats["own_goals"].fillna(0)
defensive_stats["last_man_tackles"] = defensive_stats["last_man_tackles"].fillna(0)
```
